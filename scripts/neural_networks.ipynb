Exercícios dos slides 4 e 5

```python
import numpy as np

# uma amostra com 2 features --> shape (1, 2) (2D: linhas = amostras, colunas = features)
x = np.array([[1, 2]])

# pesos da 1.ª camada (2 entradas x 3 neurónios)
# linha i = pesos que saem da entrada i; coluna j = pesos que entram no neurónio j
W = np.array([[0.1, 0.2, -0.1],
              [0.3, -0.2, 0.4]])

# um bias por neurónio --> shape (1, 3)
b = np.array([[0.1, 0, -0.1]])

# Y = X·W + B: (1, 2) · (2, 3) --> (1, 3), um valor por neurónio
print(np.dot(x, W) + b)
```

```python
# confirmação para o neurónio 1 à mão: usa a coluna 0 de W
h1 = x[0, 0] * W[0, 0] + x[0, 1] * W[1, 0] + b[0, 0]
print(round(h1, 2))  # 1*0.1 + 2*0.3 + 0.1 = 0.8
```

**Conclusão (Exercício 1):** o `np.dot(x, W) + b` dá **[0.8, −0.2, 0.6]**, igual às contas à mão. O `np.dot` faz de uma vez a soma pesada de todos os neurónios: as shapes (1, 2) · (2, 3) dão (1, 3), porque o 2 do meio (número de entradas) desaparece na soma.

Valores são a parte linear (soma pesada + bias), nao têm função de ativação. 

Exercicio 2 - Rede 2 --> 3 --> 1 dados aleatórios

X aleatório com 5 amostras e 2 features; pesos aleatórios em [−0.5, 0.5[ e biases a zero.

```python
np.random.seed(42)  # semente fixa: nº aleatórios são sempre os mesmos (resultados reprodutíveis)

# 5 amostras, 2 features, valores em [0, 1[
X = np.random.rand(5, 2)

# camada 1: 2 --> 3
# np.random.rand dá [0, 1[; subtrair 0.5 centra em zero --> [-0.5, 0.5[
W1 = np.random.rand(2, 3) - 0.5
b1 = np.zeros((1, 3))

# camada 2: 3 -> 1
W2 = np.random.rand(3, 1) - 0.5
b2 = np.zeros((1, 1))

# forward propagation: o output de uma camada é o input da seguinte
H1 = np.dot(X, W1) + b1   # (5, 2) · (2, 3) -> (5, 3)
Y = np.dot(H1, W2) + b2   # (5, 3) · (3, 1) --> (5, 1)

print("X: ", X.shape)
print("W1:", W1.shape, " b1:", b1.shape)
print("H1:", H1.shape)
print("W2:", W2.shape, " b2:", b2.shape)
print("Y: ", Y.shape)
print(Y)
```

**Conclusão (Exercício 2):**

- O número de amostras (5) mantém-se em todas as camadas; só muda o número de colunas: 2 → 3 → 1.
- Para o `np.dot` funcione: as colunas do input têm de ser iguais às linhas de W (2 = 2 na 1.ª camada, 3 = 3 na 2.ª). Por isso W tem shape (n_features, n_units).
- O bias tem shape (1, n_units) e é somado às 5 linhas por *broadcasting*: um bias por neurónio, igual para todas as amostras.
- O resultado final tem shape (5, 1): uma previsão por amostra.

## DenseLayer (slides 8–13)

Implementada em `src/si/neural_networks/layers.py`, junto com a classe base abstrata `Layer`.

```python
from si.neural_networks.layers import DenseLayer

# testes doslide 13
layer = DenseLayer(4, (3,)).initialize()                 # 4 neurónios, input com 3 features
out = layer.forward_propagation(np.random.rand(10, 3))   # 10 amostras

assert out.shape == (10, 4)               # (amostras, neurónios)
assert layer.parameters() == 3 * 4 + 4    # 12 pesos + 4 biases = 16
assert layer.output_shape() == (4,)
print("Testes do slide 13: OK")
print(layer.layer_name(), "| input_shape:", layer.input_shape(),
      "| output_shape:", layer.output_shape(), "| parâmetros:", layer.parameters())
```

```python
# refazer o Exercício 2 com a DenseLayer: as camadas recebem os mesmos pesos (W1, W2), por isso o resultado tem de ser igual ao calculado com np.dot
layer1 = DenseLayer(3, input_shape=(2,)).initialize()
layer2 = DenseLayer(1, input_shape=layer1.output_shape()).initialize()  # input da 2.ª = output da 1.ª

layer1.weights = W1
layer2.weights = W2

Y_dense = layer2.forward_propagation(layer1.forward_propagation(X))

print(Y_dense.shape) # (5, 1)
print(np.allclose(Y_dense, Y)) # True
print("Parâmetros:", layer1.parameters(), "+", layer2.parameters(), "=",
      layer1.parameters() + layer2.parameters())
```

**Conclusão (DenseLayer):** a `DenseLayer` faz o cálculo dos exercícios, Y = X·W + B, e dá o mesmo resultado que o Exercício 2 quando usa os mesmos pesos.

- **Parâmetros** = pesos + biases = n_features · n_units + n_units. Na rede 2 --> 3 --> 1: (2·3 + 3) + (3·1 + 1) = 9 + 4 = **13**.
- O `input_shape` de uma camada é o `output_shape` da anterior; é assim que a futura classe `NeuralNetwork` vai ligar as camadas.
- O `forward_propagation` guarda o `input` porque o *backward propagation* vai precisar dele para calcular o gradiente dos pesos.
