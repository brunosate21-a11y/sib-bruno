```python
# imports
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.logistic_regression import LogisticRegression
```

```python
# carregar o dataset breast-bin.csv
# features=False: o ficheiro não tem nomes de colunas
# label=True: a última coluna é a classe a prever (0 = benigno, 1 = maligno)
breast = read_csv('../datasets/breast_bin/breast-bin.csv', sep=',', features=False, label=True)
print(breast.shape())   # (amostras, features)
```

```python
# dividir em treino (80%) e teste (20%)
# random_state fixo: a divisão é sempre igual, por isso os resultados são reprodutíveis
train, test = train_test_split(breast, test_size=0.2, random_state=42)
print(train.shape())  # amostras de treino
print(test.shape())   # amostras de teste
```

```python
# criar o modelo e treiná-lo SÓ com os dados de treino
model = LogisticRegression(l2_penalty=1, alpha=0.001, max_iter=2000, patience=10, scale=True)
model.fit(train)

# avaliar no conjunto de teste (dados que o modelo nunca viu)
print(f"Score (accuracy) no teste: {model.score(test):.4f}")  # proporção de acertos: quanto mais alto, melhor
print(f"Custo no teste: {model.cost(test):.4f}")              # log-loss + penalização L2: quanto mais baixo, melhor
print(f"Iterações: {len(model.cost_history)}")                # se for igual a max_iter, não chegou a convergir
```

   **Score (accuracy) no teste:** 0.9856 — 137 de 139 amostras bem classificadas (baseline "sempre benigno": ~66%).
   **Custo no teste:** 0.1859 — log-loss + L2; o custo inicial (thetas = 0) era 0.693 = -log(0.5).
   O treino parou nas 2000 iterações (max_iter) sem convergir: com alpha = 0.001 os passos são pequenos.
